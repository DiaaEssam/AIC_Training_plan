## 1. Imports and Model Loading

```python
import spacy
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

nlp = spacy.load('en_core_web_md')
```

## 2. Corpus Processing and Vector Inspection

```python
corpus = [
    'Speech recognition converts speech into text.',
    'Natural language processing analyzes text.',
    'Machine learning improves speech recognition.',
    'Audio signals contain useful information.',
    'Word embeddings capture semantic meaning.'
]

docs = [nlp(sentence) for sentence in corpus]

for doc in docs:
    print('=' * 80)
    print(f'Document: {doc.text}')
    for token in doc:
        if token.has_vector:
            print(f'  {token.text:15} | Vector Shape: {token.vector.shape}')
```

## 3. Word Similarity Analysis

```python
pairs = [
    ('speech', 'language'),
    ('speech', 'audio'),
    ('speech', 'banana'),
    ('text', 'document'),
    ('machine', 'learning')
]

for w1, w2 in pairs:
    sim = nlp(w1).similarity(nlp(w2))
    print(f'{w1:10} ↔ {w2:10} : {sim:.3f}')
```

## 4. Sentence and Document Similarity

```python
doc1 = nlp('Speech recognition converts audio into text.')
doc2 = nlp('Audio transcription converts spoken language into words.')
doc3 = nlp('I enjoy eating pizza.')

print(f'Similarity (Doc1 ↔ Doc2): {doc1.similarity(doc2):.3f}')
print(f'Similarity (Doc1 ↔ Doc3): {doc1.similarity(doc3):.3f}')
```

## 5. Finding Most Similar Words (Vocab Search)

```python
query = 'speech'
queries = np.asarray([nlp.vocab[query].vector])

ids, _, scores = nlp.vocab.vectors.most_similar(queries, n=10)

print(f'Words most similar to "{query}":\n')
for word_id, score in zip(ids[0], scores[0]):
    print(f'{nlp.vocab.strings[word_id]:15} {score:.3f}')
```

## 6. PCA Projection (Dimensionality Reduction)

```python
words = ['speech', 'audio', 'voice', 'recognition', 'language', 'text', 'machine', 'learning', 'music', 'banana']
vectors = np.array([nlp(word).vector for word in words])

pca = PCA(n_components=2)
points = pca.fit_transform(vectors)

plt.figure(figsize=(10, 8))
for word, point in zip(words, points):
    plt.scatter(point[0], point[1], s=100, color='dodgerblue')
    plt.text(point[0] + 0.02, point[1] + 0.02, word, fontsize=12)

plt.title('2D Projection of Word Embeddings (PCA)', fontsize=15)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.grid(True, alpha=0.3)
plt.show()
```
